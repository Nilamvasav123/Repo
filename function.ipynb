```python
'''Problem 6: Grade Counter

Design a function that:
	•	Accepts a list of marks
	•	Uses a loop and conditions to count:
	•	Fail (<40)
	•	Pass (40–59)
	•	Distinction (≥60)
	•	Return all three counts'''

def grade1(*counter):
    fail_cnt=0
    pass_cnt=0
    dist_cnt=0
    for mk in counter:
        if mk < 40:
            fail_cnt +=1
            
        elif 40<= mk <=59:
            pass_cnt +=1
            
        elif mk >=60:
            dist_cnt +=1
            
    return  fail_cnt,pass_cnt,dist_cnt

print(grade1(30,45,70,79))  
```

```python
'''Problem 9: Duplicate Finder

Create a function that:
	•	Loops through a list
	•	Returns the first duplicate element
	•	If no duplicate exists, return “No Duplicates”'''

def dublicate1(*finder):
    count=0
    for i in finder:
        count +=1 # Compare current element with all previous ones
        for j in range(count - 1):
            if i == finder[j]:
                return i  # First duplicate found
    return "no dublicates"


print(dublicate1(1, 2, 2, 3))   # Output: 2
print(dublicate1(1, 2, 3, 4))   # Output: no dublicates

```

```python
'''Problem 11: Consecutive Failure Detector

Write a function that:
	•	Accepts a list of 0s and 1s (1 = failure)
	•	Uses loop to detect:
	•	Return “Critical” if 3 consecutive failures occur
	•	Else return “Normal”'''

def consecutive_failure_detector(events):
    count = 0
    for e in events:
        if e == 1:  # failure
            count += 1
            if count == 3:
                return "Critical"
        else:
            count = 0  # reset if not a failure
    return "Normal"

print(consecutive_failure_detector([0, 1, 1, 1, 0])) # Critical 
print(consecutive_failure_detector([1, 0, 1, 1, 0])) # Normal 
print(consecutive_failure_detector([0, 0, 0])) 
# Normal
print(consecutive_failure_detector([1, 1, 1, 1]))


```

```python
'''Problem 15: Order Fulfillment

Create a function that:
	•	Takes list of order quantities
	•	Uses loop to deduct from stock
	•	If stock becomes negative → return “Insufficient Stock”
	•	Else return remaining stock'''

def fulfill_order(order_quantities, stock):
    """
    Deducts order quantities from stock.
    
    Parameters:
        order_quantities (list): List of integers representing order quantities.
        stock (int): Initial stock available.
    
    Returns:
        str or int: "Insufficient Stock" if stock goes negative,
                    otherwise the remaining stock.
    """
    for qty in order_quantities:
        stock -= qty
        if stock < 0:
            return "Insufficient Stock"
    return stock


# Example usage:
print(fulfill_order([2, 3, 4], 10))   # Output: 1
print(fulfill_order([5, 7], 10))      # Output: "Insufficient Stock"


```
