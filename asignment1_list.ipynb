```python
'''13.	Find second highest number
Input: list of ints
Task: find second max using loop (no sort), handle duplicates.'''

def find_second_max(lst):
    # सुरुवातीला दोन variables ठेवूया
    first_max = float('-inf')   # सर्वात मोठा number
    second_max = float('-inf')  # दुसरा मोठा number

    for num in lst:
        # जर num हा first_max पेक्षा मोठा असेल
        if num > first_max:
            second_max = first_max
            first_max = num
        # जर num हा first_max पेक्षा लहान पण second_max पेक्षा मोठा असेल
        elif num > second_max and num != first_max:
            second_max = num

    # जर second_max अजूनही -inf असेल म्हणजे दुसरा max नाही
    return second_max if second_max != float('-inf') else None


# Example usage:
numbers = [20, 40, 90, 80, 50]
print(find_second_max(numbers))  # Output: 80

numbers2 = [7, 7, 7]
print(find_second_max(numbers2))  # Output: None (कारण दुसरा max नाही)

```

```python
'''37.	Rotate list right by k
Input list, k
Output rotated list using slicing or loop.'''

def rotate_list(lst, k):
    k = k % len(lst)  # handle cases where k > len(lst)
    return lst[-k:] + lst[:-k]

# Example
print(rotate_list([1, 2, 3, 4, 5], 2))  # Output: [4, 5, 1, 2, 3]


#loop
def rotate_list_loop(lst, k):
    k = k % len(lst)
    for _ in range(k):
        last = lst.pop()   # remove last element
        lst.insert(0, last)  # insert at front
    return lst

# Example
print(rotate_list_loop([1, 2, 3, 4, 5], 2))  # Output: [4, 5, 1, 2, 3]

```

```python
'''38.	Sort only even numbers, keep odds in place
Example: [5,2,8,1,4]
Evens sorted -> [4,8,2]? (Students must keep odd positions unchanged.)'''


def sort_evens_keep_odds(arr):
    evens = sorted([x for x in arr if x % 2 == 0])
    result = []
    even_index = 0
    for num in arr:
        if num % 2 == 0:
            result.append(evens[even_index])
            even_index += 1
        else:
            result.append(num)
    return result

# Example
print(sort_evens_keep_odds([5, 2, 8, 1, 4]))  # Output: [5, 2, 4, 1, 8]
```

```python
'''56.	Sessionization (simplified)
Input: list of event times (minutes) sorted
Rule: new session if gap > 30
Output: list of sessions as list of lists.'''

# Input: sorted list of event times (minutes)
events = [5, 10, 20, 55, 65, 120]

sessions = []              # to store grouped sessions
current_session = [events[0]]  # start with the first event

for i in range(1, len(events)):
    if events[i] - events[i-1] > 30:
        # gap > 30 → start a new session
        sessions.append(current_session)
        current_session = [events[i]]
    else:
        # same session
        current_session.append(events[i])

# add the last session
sessions.append(current_session)

print("Sessions:", sessions)
```

```python
'''57.	Split logs by severity
Input: logs like "2026-01-01|ERROR|msg"
Task: create 3 lists: errors, warns, infos (parse using split).'''

# Input logs
logs = [
    "2026-01-01|ERROR|Something failed",
    "2026-01-02|WARN|Check configuration",
    "2026-01-03|INFO|System started",
    "2026-01-04|ERROR|Disk full",
    "2026-01-05|INFO|User logged in"
]

# Lists to hold different severities
errors = []
warns = []
infos = []

# Process each log
for log in logs:
    parts = log.split("|")   # split into [date, severity, message]
    severity = parts[1]
    message = parts[2]

    if severity == "ERROR":
        errors.append(message)
    elif severity == "WARN":
        warns.append(message)
    elif severity == "INFO":
        infos.append(message)

print("Errors:", errors)
print("Warns:", warns)
print("Infos:", infos)
```

```python
'''60.	Build a simple leaderboard
Input: player list, score list
Combine, sort desc, output top N, also show lowest player.'''


players = ["nilam", "priyanka"]
scores = [100, 200]

# Combine players and scores
leaderboard = list(zip(players, scores))

# Sort by score descending
leaderboard.sort(key=lambda x: x[1], reverse=True)

# Show full leaderboard
print("Leaderboard:", leaderboard)

# Top N (example: Top 1)
N = 1
print("Top", N, ":", leaderboard[:N])

# Lowest player
print("Lowest player:", leaderboard[-1])

'''🔥 Full Flow
zip → combine
sort → मोठा score वर
[:N] → top players
[-1] → lowest playe'''
```

```python

'''62.	Keep only palindromes from list of strings.'''

words = ["madam", "apple", "racecar", "banana", "level", "python"]
palindromes = []

for w in words:
    if w == w[::-1]:   # check if word equals its reverse
        palindromes.append(w)

print(palindromes)

```
