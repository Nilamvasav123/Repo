```python
'''1.	Clean API ages
Input: ages = ["25", "30", "", None, "18", " 40 "]
Task: remove invalid values ("", None, non-digits), strip spaces, convert to int, print len/min/max/sum.'''

# Input list
ages = ["25", "30", "", None, "18", " 40 "]

# Clean the list: remove invalids, strip spaces, keep only digits, convert to int
cleaned_ages = [int(age.strip()) for age in ages if age and age.strip().isdigit()]

# Print required values
print("Cleaned Ages:", cleaned_ages)
print("Length:", len(cleaned_ages))
print("Min:", min(cleaned_ages))
print("Max:", max(cleaned_ages))
print("Sum:", sum(cleaned_ages))
```

```python
'''3.	Validate all emails contain “@”
Input: list of emails
Task: using loop + condition build list of invalid emails; print any(invalid) and all(valid'''

emails = ["user@example.com", "testmail.com", "hello@world.org", "noatsymbol"]

invalid = []
for email in emails:
    if "@" not in email:
        invalid.append(email)

print("Invalid emails:", invalid)
print("Any invalid?", any(invalid))   # True if list is non-empty
print("All valid?", all(invalid))
```

```python
'''4.	Remove negative prices
Input: prices = [100, -5, 200, 0, 50]
Task: keep only >0, then insert missing default price 99 at index 0 if list becomes empty.'''


prices = [100, -5, 200, 0, 50]
result = [i for i in prices if i > 0]
if not result:
    result.insert(0, 99)

print(result)
```

```python
'''8.	Split valid/invalid based on rule
Input: records = ["ID-100", "ID-2", "ABC", "ID-999"]
Rule: valid if starts with "ID-" and last part is digits. Separate two lists.'''

records = ["ID-100", "ID-2", "ABC", "ID-999"]

valid = []
invalid = []

for r in records:
    if r.startswith("ID-") and r[3:].isdigit():
        valid.append(r)
    else:
        invalid.append(r)

print("Valid:", valid)
print("Invalid:", invalid)



```

```python
'''11.	Find first occurrence index (manual)
Input: list and target
Task: return first index using loop; if not found return -1 (don’t use .index()).'''"data_type practice.ipynb"

def find_first_index(lst, target):
    # enumerate gives both index and value directly
    for i, val in enumerate(lst):
        if val == target:
            return i   # Return the first matching index
    return -1          # If not found, return -1

# Example usage:
numbers = [10, 20, 30, 40, 20, 50]
print(find_first_index(numbers, 20))  # Output: 1
print(find_first_index(numbers, 60))  # Output: -1
```

```python
'''13.	Find second highest number
Input: list of ints
Task: find second max using loop (no sort), handle duplicates.'''

def find_second_max(lst):
    # सुरुवातीला दोन variables ठेवूया
    first_max = float('-inf')   # सर्वात मोठा number
    second_max = float('-inf')  # दुसरा मोठा number

    for num in lst:
        # जर num हा first_max पेक्षा मोठा असेल
        if num > first_max:
            second_max = first_max
            first_max = num
        # जर num हा first_max पेक्षा लहान पण second_max पेक्षा मोठा असेल
        elif num > second_max and num != first_max:
            second_max = num

    # जर second_max अजूनही -inf असेल म्हणजे दुसरा max नाही
    return second_max if second_max != float('-inf') else None


# Example usage:
numbers = [20, 40, 90, 80, 50]
print(find_second_max(numbers))  # Output: 80

numbers2 = [7, 7, 7]
print(find_second_max(numbers2))  # Output: None (कारण दुसरा max नाही)

```

```python
'''14.	Find longest word and its index
Input: list of words
Task: return word + index.'''

names=["Gaurav","priyanka","Nilam"]

# Find the longest word and its index
longest_word = max(names, key=len)        # longest word by length
index = names.index(longest_word)         # index of that word

print("Longest word:", longest_word)
print("Index:", index)
```
