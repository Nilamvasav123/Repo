```python
'''9. Resouce limitation 
    rule:
    - Total resouces = 100 
    - consume values one by one 
    - If resource goes negative -> STOP 

    requests= [20,30,40,25]
    resouces = 100 '''

'''rts = [20,30,40,25]
re = 100
for r in rts:
    if r-=re:
       print("you have enough money")
    else:
        print("you dont have enough money")'''


requests = [20, 30, 40, 25]
resources = 100

for req in requests:
    if resources - req < 0:
        print("STOP")
        break
    else:
        resources -= req
        print(f"Consumed {req}, Remaining resources = {resources}")
else:
    print("All requests processed successfully")
```

```python
'''11. Budget Control 
    budget=150 
    expense=[30,40,60,50]

    Rule: 
    - Subtract expense one by one 
    - If budget becomes negative -> STOP 
    - Peint remaning budget after each valid expense'''




budget = 150
expenses = [30, 40, 60, 50]

for expense in expenses:
    budget -= expense
    if budget < 0:
        print("Budget went negative! Stopping...")
        break
    else:
        print("Remaining budget:", budget)
```

```python
'''13. levels = [100,90,80,85,70]
    Rules :
        - If values drop consecutively:
            - if 3 continious drop -> Critical 
            - Else - Warning 
        - Else -> NORMAL 
'''



levels = [100, 90, 80, 85, 70]

drop_count = 0

for i in range(1, len(levels)):
    if levels[i] < levels[i-1]:
        drop_count += 1

if drop_count >= 3:
    print("Critical")
elif drop_count > 0:
    print("Warning")
else:
    print("Normal")

```
