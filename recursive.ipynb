```python
#1. Print Number N-1 

def number(n):
    if n==0:
        return 
    print(n)
    return number(n-1)

number(5)
```

```python
#2. Print Number 1 - N 

def num(n):
    if n==0:
        return
    num(n-1)
    print(n)

num(5)
```

```python
#3. Sum of First N numbers 

def total(n):
    if n==0:
        return 0
    return n + total(n-1)

total(5)
```

```python
#4. Factorial 
#5!=5!*4!*3!*2!*1! =120

def fact(n):
    if n==0 or n==1:
        return 1
    return n* fact(n-1)

fact(5)
```

```python
#5. Count Digits 

def count_digits(n):
    if n==0:
        return 0
    return 1+count_digits(n//10)
print(count_digits(12345))
```

```python
'''6. Sum of Digits = Input 1234 , Output : 10 
'''

def sum_of_digits(n):
    if n == 0:              # base case
        return 0
    return (n % 10) + sum_of_digits(n // 10)

sum_of_digits(1234)
```

```python
'''7. Power of (x^n) = Input x=2, n=3 , Output : 8 '''
def power(x,n):
    if n==0:
        return 1
    return x *power(x,n-1)

power(2, 3)
```
